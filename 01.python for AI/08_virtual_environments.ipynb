#  Why Virtual Environments?

Suppose Project A requires:

    numpy 1.x

while Project B requires:

    numpy 2.x

Installing everything globally can create conflicts.

A virtual environment gives each project its own isolated Python environment.


#  What is .venv?

`.venv` is a directory containing an isolated Python environment for a project.

The `.venv` directory contains:

- Python environment information
- installed packages
- environment-specific executables

We normally DO NOT upload `.venv` to GitHub.

#  Creating a Virtual Environment

Open PowerShell or Command Prompt.

First, navigate to your project folder:

cd "path\to\python for ai"

Then create the environment:

python -m venv .venv


## PowerShell

.venv\Scripts\Activate.ps1

## Command Prompt (CMD)

.venv\Scripts\activate

After activation, your terminal usually shows:

(.venv) ...

### Environment Diagnostic Tool

```python
import sys
import importlib.util

packages = [
    "numpy",
    "pandas",
    "matplotlib",
    "requests",
    "torch",
    "transformers"
]

print("Python:")
print(sys.executable)

print("\nPackage availability:")

for package in packages:
    installed = importlib.util.find_spec(package) is not None
    status = "Installed" if installed else "Not installed"
    print(f"{package:15} -> {status}")
```
