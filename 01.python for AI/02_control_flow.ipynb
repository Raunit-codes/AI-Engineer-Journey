```python
age = 20

if age >= 18:
    print("You are an adult.")
```

```python
score = 85

if score >= 50:
    print("Student passed.")
```

```python
marks = 78

if marks >= 90:
    grade = "A+"
elif marks >= 80:
    grade = "A"
elif marks >= 70:
    grade = "B"
elif marks >= 60:
    grade = "C"
elif marks >= 50:
    grade = "D"
else:
    grade = "F"

print("Grade:", grade)
```

```python
age = 22
has_id = True

if age >= 18 and has_id:
    print("Access granted.")
else:
    print("Access denied.")
```

```python
is_admin = False
is_manager = True

if is_admin or is_manager:
    print("You have administrative access.")
else:
    print("Access denied.")
```

```python
is_logged_in = False

if not is_logged_in:
    print("Please log in.")
```

```python
age = 20
has_id = True

if age >= 18:
    if has_id:
        print("Entry allowed.")
    else:
        print("ID required.")
else:
    print("You must be 18 or older.")
```

```python
languages = ["Python", "C++", "Java"]

for language in languages:
    print(language)
```

```python
for i in range(5):
    print(i)
```

```python
for i in range(0, 11, 2):
    print(i)
```

```python
total = 0

for i in range(1, 11):
    total += i

print("Total:", total)
```

```python
marks = [78, 92, 45, 67, 88]

for mark in marks:
    if mark >= 50:
        print(mark, "Pass")
    else:
        print(mark, "Fail")
```

find max

```python
numbers = [23, 67, 12, 89, 45]

maximum = numbers[0]

for number in numbers:
    if number > maximum:
        maximum = number

print("Maximum:", maximum)
```

```python
count = 1

while count <= 5:
    print(count)
    count += 1
```

```python
for i in range(1, 11):

    if i == 6:
        break

    print(i)
```

```python
for i in range(1, 11):

    if i == 5:
        continue

    print(i)
```

pass does nothing.

It's mainly useful as a placeholder when you're writing code that you'll implement later.

```python
for i in range(5):

    if i == 2:
        pass

    print(i)
```

```python
for i in range(3):

    for j in range(3):
        print(i, j)
```

```python
for i in range(1, 6):
    print("*" * i)
```

AI-Related Example

```python
score = 0.87

if score >= 0.90:
    prediction = "High confidence"
elif score >= 0.70:
    prediction = "Medium confidence"
else:
    prediction = "Low confidence"

print("Model confidence:", prediction)
```

```python
temperature = 38.5

if temperature >= 40:
    status = "Very High"
elif temperature >= 35:
    status = "High"
elif temperature >= 25:
    status = "Normal"
else:
    status = "Low"

print("Temperature status:", status)
```

```python
predictions = [0.91, 0.72, 0.43, 0.88, 0.31]

for probability in predictions:

    if probability >= 0.80:
        print(probability, "→ Positive")
    else:
        print(probability, "→ Negative")
```

```python
students = {
    "Aman": 85,
    "Riya": 72,
    "Karan": 45,
    "Neha": 91,
    "Arjun": 38
}

passed = 0
failed = 0

for name, marks in students.items():

    if marks >= 90:
        grade = "A+"
    elif marks >= 80:
        grade = "A"
    elif marks >= 70:
        grade = "B"
    elif marks >= 60:
        grade = "C"
    elif marks >= 50:
        grade = "D"
    else:
        grade = "F"

    if marks >= 50:
        status = "Pass"
        passed += 1
    else:
        status = "Fail"
        failed += 1

    print(f"{name}: {marks} | Grade: {grade} | {status}")

print("\nSummary")
print("Passed:", passed)
print("Failed:", failed)
```
