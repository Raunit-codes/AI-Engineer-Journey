##  What is Object-Oriented Programming?

OOP organizes code around objects.

An object can contain:

- Data → attributes
- Behavior → methods

A class is a blueprint used to create objects.

Example:

Class:
    Student

Objects:
    student1
    student2
    student3

```python
class Student:
    pass
```

```python
student1 = Student()

print(student1)
print(type(student1))
```

```python
class Student:
    pass


student1 = Student()

student1.name = "Ram"
student1.age = 21
student1.branch = "AI"

print(student1.name)
print(student1.age)
print(student1.branch)
```

##  The `__init__()` Constructor

`__init__()` is automatically called when an object is created.

It is commonly used to initialize object attributes.

```python
class Student:

    def __init__(self, name, age, branch):
        self.name = name
        self.age = age
        self.branch = branch
        
```

```python
student1 = Student("Ray", 19, "AI")
student2 = Student("Aman", 20, "CSE")

print(student1.name)
print(student2.name)
```

##  Methods

A method is a function defined inside a class.
Methods describe the behavior of an object.

```python
class Student:

    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    def display(self):
        print(f"Name: {self.name}")
        print(f"Marks: {self.marks}")


student1 = Student("Rahul", 85)

student1.display()
```

```python
class Student:

    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    def is_passed(self):
        return self.marks >= 50


student = Student("Rahul", 95)

print(student.is_passed())
```

```python
class Student:

    def __init__(self, name, marks):
        self.name = name
        self.marks = marks


student = Student("Rahul", 70)

print(student.marks)

student.marks = 90

print(student.marks)
```

##  Encapsulation

Encapsulation means keeping data and the methods that operate
on that data together inside a class.

Python uses naming conventions to indicate that an attribute
is intended to be internal/private.

```python
class BankAccount:

    def __init__(self, balance):
        self._balance = balance
```

```python
Account1  = BankAccount(10000)
print(Account1._balance)
```

! The _balance naming convention indicates that this attribute is intended for internal use.
It is not a strict access restriction.

```python
class BankAccount:

    def __init__(self, balance):
        self.__balance = balance

    def get_balance(self):
        return self.__balance


account = BankAccount(10000)

print(account.get_balance())
```

##  Inheritance

Inheritance allows one class to reuse functionality from another class.

Parent class:
    Animal

Child classes:
    Dog
    Cat

```python
class Animal:

    def speak(self):
        print("Animal makes a sound")


class Dog(Animal):
    pass


dog = Dog()

dog.speak()
```

```python
class Animal:

    def speak(self):
        print("Animal makes a sound")


class Dog(Animal):

    def bark(self):
        print("Dog barks")


dog = Dog()

dog.speak()
dog.bark()
```

```python
class Animal:

    def speak(self):
        print("Animal makes a sound")


class Dog(Animal):

    def speak(self):
        print("Dog barks")


animal = Animal()
dog = Dog()

animal.speak()
dog.speak()
```

```python
class Animal:

    def __init__(self, name):
        self.name = name


class Dog(Animal):

    def __init__(self, name, breed):
        super().__init__(name)
        self.breed = breed


dog = Dog("Bruno", "Labrador")

print(dog.name)
print(dog.breed)

#super() allows the child class to call functionality from its parent class.
```

##  Polymorphism

Polymorphism means that different objects can respond to the
same method/interface in different ways.

```python
class Dog:

    def speak(self):
        print("Woof")


class Cat:

    def speak(self):
        print("Meow")


animals = [Dog(), Cat()]

for animal in animals:
    animal.speak()
```

### AI Example — Simple Classifier

```python
class Classifier:

    def __init__(self, threshold=0.5):
        self.threshold = threshold

    def predict(self, probability):

        if probability >= self.threshold:
            return "Positive"

        return "Negative"


classifier = Classifier(threshold=0.7)

print(classifier.predict(0.85))
print(classifier.predict(0.52))
```

```python
class Classifier:

    def __init__(self, threshold=0.5):
        self.threshold = threshold

    def predict(self, probability):

        if probability >= self.threshold:
            return 1

        return 0

    def predict_batch(self, probabilities):

        predictions = []

        for probability in probabilities:
            predictions.append(
                self.predict(probability)
            )

        return predictions


classifier = Classifier(threshold=0.7)

probabilities = [0.91, 0.43, 0.82, 0.61, 0.95]

predictions = classifier.predict_batch(probabilities)

print(predictions)
```

```python
class SimpleLinearModel:

    def __init__(self, name, weight, bias):
        self.name = name
        self.weight = weight
        self.bias = bias

    def predict(self, x):
        return self.weight * x + self.bias

    def predict_batch(self, values):

        predictions = []

        for value in values:
            predictions.append(
                self.predict(value)
            )

        return predictions

    def display_parameters(self):

        print("Model:", self.name)
        print("Weight:", self.weight)
        print("Bias:", self.bias)


model = SimpleLinearModel(
    name="Simple Linear Model",
    weight=2,
    bias=5
)

model.display_parameters()

print(model.predict(10))

values = [1, 2, 3, 4, 5]

print(model.predict_batch(values))
```
