##  What is Object-Oriented Programming?

OOP organizes code around objects.

An object can contain:

- Data → attributes
- Behavior → methods

A class is a blueprint used to create objects.

Example:

Class:
    Student

Objects:
    student1
    student2
    student3

```python
class Student:
    pass
```

```python
student1 = Student()

print(student1)
print(type(student1))
```

```python
class Student:
    pass


student1 = Student()

student1.name = "Ram"
student1.age = 21
student1.branch = "AI"

print(student1.name)
print(student1.age)
print(student1.branch)
```

##  The `__init__()` Constructor

`__init__()` is automatically called when an object is created.

It is commonly used to initialize object attributes.

```python
class Student:

    def __init__(self, name, age, branch):
        self.name = name
        self.age = age
        self.branch = branch
        
```

```python
student1 = Student("Ray", 19, "AI")
student2 = Student("Aman", 20, "CSE")

print(student1.name)
print(student2.name)
```
