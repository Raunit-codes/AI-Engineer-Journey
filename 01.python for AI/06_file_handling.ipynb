## Common File Modes

"r"
→ Read

"w"
→ Write
→ Creates a file or overwrites an existing file

"a"
→ Append
→ Adds data to the end

"x"
→ Create a new file
→ Raises an error if the file already exists

"b"
→ Binary mode

Example:

"rb"
→ Read binary

"wb"
→ Write binary

###
 Dynamic Example — Student Data

```python
from pathlib import Path

data_folder = Path("data")
data_folder.mkdir(exist_ok=True)

file_path = data_folder / "students.txt"

number_of_students = int(
    input("How many students? ")
)

with open(file_path, "w") as file:

    for i in range(number_of_students):

        name = input(f"Enter student {i + 1} name: ")
        marks = input(f"Enter marks for {name}: ")

        file.write(f"{name},{marks}\n")

print("Student data saved.")
```

```python
with open(file_path, "r") as file:

    for line in file:
        print(line.strip())
```

```python
students = []

with open(file_path, "r") as file:

    for line in file:

        name, marks = line.strip().split(",")

        students.append({
            "name": name,
            "marks": float(marks)
        })

print(students)
```

# CSV Files

CSV = Comma-Separated Values

Example:

name,age,marks
Aman,20,85
Riya,19,92
Karan,21,67

CSV is one of the most common formats for tabular datasets.

```python
import csv

students = [
    ["Aman", 20, 85],
    ["Riya", 19, 92],
    ["Karan", 21, 67]
]

with open("students.csv", "w", newline="") as file:

    writer = csv.writer(file)

    writer.writerow(["name", "age", "marks"])

    writer.writerows(students)

print("CSV created.")
```

```python
with open("students.csv", "r") as file:

    reader = csv.reader(file)

    for row in reader:
        print(row)
```

```python
with open("students.csv", "r") as file:

    reader = csv.DictReader(file)

    for row in reader:
        print(row)
```

### Dynamic CSV Analyzer

```python
import csv

total_marks = 0
student_count = 0

with open("students.csv", "r") as file:

    reader = csv.DictReader(file)

    for row in reader:

        marks = float(row["marks"])

        total_marks += marks
        student_count += 1

average = total_marks / student_count

print("Students:", student_count)
print("Average marks:", average)
```

# JSON

JSON is extremely common when working with APIs.

Example:

{
    "name": "Ray",
    "age": 19,
    "skills": ["Python", "SQL", "AI"]
}

Python dictionaries and JSON objects have a very similar structure.

```python
import json

student = {
    "name": "Ram",
    "age": 19,
    "branch": "AI",
    "skills": [
        "Python",
        "SQL",
        "Machine Learning"
    ]
}

with open("student.json", "w") as file:

    json.dump(
        student,
        file,
        indent=4
    )

print("JSON file created.")
```

```python
with open("student.json", "r") as file:

    student = json.load(file)

print(student)
```

```python
### Dynamic JSON Example
```

```python
name = input("Enter your name: ")
age = int(input("Enter your age: "))
branch = input("Enter your branch: ")

skills = input(
    "Enter your skills separated by commas: "
).split(",")

profile = {
    "name": name,
    "age": age,
    "branch": branch,
    "skills": [skill.strip() for skill in skills]
}

with open("profile.json", "w") as file:

    json.dump(
        profile,
        file,
        indent=4
    )

print("Profile saved.")
```

```python
with open("profile.json", "r") as file:
    profile = json.load(file)

print("\nSaved Profile")

for key, value in profile.items():
    print(f"{key}: {value}")
```
