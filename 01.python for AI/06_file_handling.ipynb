```python
file = open("sample.txt", "w")

file.write("Hello from Python!")

file.close()
```

```python
with open("sample.txt", "w") as file:
    file.write("Hello from Python!")

print("File written successfully.")
```

```python
with open("sample.txt", "r") as file:
    content = file.read()

print(content)
```

"r" → read
"w" → write
"a" → append

```python
name = input("Enter Your Name")

with open("user.txt","w") as file:
    file.write(name)

print("Name saved successfully")
```

```python
with open("user.txt","r") as file:
    file.read()

print("Saved name:", name)
```

```python
skills = [
    "Python",
    "NumPy",
    "pandas",
    "Machine Learning",
    "Deep Learning"
]

with open("skills.txt", "w") as file:

    for skill in skills:
        file.write(skill + "\n")

print("Skills saved.")
```

```python
with open("skills.txt", "r") as file:
    lines = file.readlines()

print(lines)
```

```python
with open("skills.txt", "r") as file:
    lines = [line.strip() for line in file]

print(lines)
```

```python
with open("skills.txt", "r") as file:

    for line in file:
        print(line.strip())
```

```python
with open("skills.txt", "a") as file:
    file.write("Computer Vision\n")
```

```python
with open("skills.txt", "r") as file:
    print(file.read())
```

```python
note = input("Enter a note: ")

with open("notes.txt", "a") as file:
    file.write(note + "\n")

print("Note saved.")
```

```python
with open("notes.txt", "r") as file:

    notes = file.readlines()

print("Your notes:")

for index, note in enumerate(notes, start=1):
    print(f"{index}. {note.strip()}")
```

## Common File Modes

"r"
→ Read

"w"
→ Write
→ Creates a file or overwrites an existing file

"a"
→ Append
→ Adds data to the end

"x"
→ Create a new file
→ Raises an error if the file already exists

"b"
→ Binary mode

Example:

"rb"
→ Read binary

"wb"
→ Write binary

Creating a Directory

```python
data_folder = Path("data")
```

```python
data_folder.mkdir(exist_ok=True)

print("Data folder ready.")
```

Writing to a Path

```python
file_path = data_folder / "students.txt"
```

```python
file_path.write_text(
    "Aman,85\nRiya,92\nKaran,67\n"
)

print("File created:", file_path)
```

```python
content = file_path.read_text()

print(content)
```
