## else

The `else` block executes only when the `try` block
finishes successfully.

```python
try:
    number = int(input("Enter a number: "))

except ValueError:
    print("Invalid input.")

else:
    print("Successfully converted:", number)
```

```python
try:
    number = int(input("Enter a number: "))

except ValueError:
    print("Invalid input.")

finally:
    print("This always executes.")
```

try + except + else + finally

```python
try:
    number = int(input("Enter a number: "))
    result = 100 / number

except ValueError:
    print("Input must be an integer.")

except ZeroDivisionError:
    print("Cannot divide by zero.")

else:
    print("Result:", result)

finally:
    print("Operation finished.")
    
```

```python
def calculator():

    try:
        first = float(input("Enter first number: "))
        operator = input("Enter operator (+, -, *, /): ")
        second = float(input("Enter second number: "))

        if operator == "+":
            result = first + second

        elif operator == "-":
            result = first - second

        elif operator == "*":
            result = first * second

        elif operator == "/":
            result = first / second

        else:
            print("Unknown operator.")
            return

        print("Result:", result)

    except ValueError:
        print("Please enter valid numbers.")

    except ZeroDivisionError:
        print("Cannot divide by zero.")


calculator()
```

```python
def calculate_percentage(marks, total):

    if total <= 0:
        raise ValueError("Total marks must be greater than zero.")

    if marks < 0:
        raise ValueError("Marks cannot be negative.")

    if marks > total:
        raise ValueError("Marks cannot exceed total marks.")

    return (marks / total) * 100


print(calculate_percentage(85, 100))
```

### Custom Exceptions

```python
class InvalidMarksError(Exception):
    pass
```

```python
def validate_marks(marks):

    if marks < 0 or marks > 100:
        raise InvalidMarksError(
            "Marks must be between 0 and 100."
        )

    return True


try:
    marks = float(input("Enter marks: "))

    validate_marks(marks)

    print("Valid marks.")

except InvalidMarksError as error:
    print("Validation error:", error)

except ValueError:
    print("Please enter a number.")
```

### Dynamic File Reader

```python
filename = input("Enter filename: ")

try:

    with open(filename, "r") as file:
        content = file.read()

    print("\nFile contents:")
    print(content)

except FileNotFoundError:
    print(f"Could not find '{filename}'.")

except PermissionError:
    print("You do not have permission to read this file.")
```

### Handling JSON Errors

```python
import json

filename = input("Enter JSON filename: ")

try:

    with open(filename, "r") as file:
        data = json.load(file)

    print("JSON loaded successfully.")
    print(data)

except FileNotFoundError:
    print("File does not exist.")

except json.JSONDecodeError:
    print("The file contains invalid JSON.")
```

### Dynamic Data validation

```python
def get_positive_number():
    while True:
        try:
            value = float(
                input("Enter a positive number:")
            )
            if value <= 0:
                raise ValueError(
                    "Number must be positive"
                )
            return value
        except ValueError as error:
            print("Invalid input:", error)

number = get_positive_number()
print("Accepted value:",number)
```

### Dynamic Student Entry

```python
def get_marks():

    while True:

        try:
            marks = float(
                input("Enter marks (0-100): ")
            )

            if not 0 <= marks <= 100:
                raise ValueError(
                    "Marks must be between 0 and 100."
                )

            return marks

        except ValueError as error:
            print("Invalid:", error)


name = input("Enter student name: ")
marks = get_marks()

student = {
    "name": name,
    "marks": marks
}

print("\nStudent record:")
print(student)
```

### Processing a Dataset Safely

```python
raw_marks = [
    "85",
    "92",
    "invalid",
    "76",
    "N/A",
    "88"
]

valid_marks = []

for value in raw_marks:

    try:
        marks = float(value)
        valid_marks.append(marks)

    except ValueError:
        print(f"Skipping invalid value: {value}")


print("Valid marks:", valid_marks)
```

### Calculating Statistics Safely

```python
raw_marks = [
    "85",
    "92",
    "invalid",
    "76",
    "N/A",
    "88"
]

valid_marks = []

for value in raw_marks:

    try:
        valid_marks.append(float(value))

    except ValueError:
        continue


if valid_marks:

    average = sum(valid_marks) / len(valid_marks)

    print("Valid values:", len(valid_marks))
    print("Average:", average)

else:
    print("No valid numeric data found.")
```

### Practical AI Example - Model Configuration

```python
def get_learning_rate():

    while True:

        try:
            rate = float(
                input("Learning rate: ")
            )

            if rate <= 0:
                raise ValueError(
                    "Learning rate must be positive."
                )

            return rate

        except ValueError as error:
            print("Invalid:", error)


learning_rate = get_learning_rate()

print("Selected learning rate:", learning_rate)
```

```python
def get_integer(prompt, minimum=None):

    while True:

        try:
            value = int(input(prompt))

            if minimum is not None and value < minimum:
                raise ValueError(
                    f"Value must be >= {minimum}."
                )

            return value

        except ValueError as error:
            print("Invalid input:", error)


epochs = get_integer(
    "Enter number of epochs: ",
    minimum=1
)

batch_size = get_integer(
    "Enter batch size: ",
    minimum=1
)

print("\nModel configuration")
print("Epochs:", epochs)
print("Batch size:", batch_size)
```

# Mini Project: Robust Dataset Cleaner

Build a program that receives raw student data containing
potentially invalid marks.

Example:

raw_students = [
    {"name": "Aman", "marks": "85"},
    {"name": "Riya", "marks": "92"},
    {"name": "Karan", "marks": "invalid"},
    {"name": "Neha", "marks": "76"},
    {"name": "Arjun", "marks": "N/A"}
]

Requirements:

1. Convert marks to float.
2. Skip invalid records.
3. Store valid records.
4. Calculate average marks.
5. Find the highest scorer.
6. Display how many records were rejected.

```python
raw_students = [
    {"name": "Aman", "marks": "85"},
    {"name": "Riya", "marks": "92"},
    {"name": "Karan", "marks": "invalid"},
    {"name": "Neha", "marks": "76"},
    {"name": "Arjun", "marks": "N/A"}
]

valid_students = []
rejected_count = 0

for student in raw_students:

    try:
        marks = float(student["marks"])

        valid_students.append({
            "name": student["name"],
            "marks": marks
        })

    except (ValueError, KeyError):
        rejected_count += 1


if valid_students:

    marks = [
        student["marks"]
        for student in valid_students
    ]

    average = sum(marks) / len(marks)

    highest = max(
        valid_students,
        key=lambda student: student["marks"]
    )

    print("Valid students:", valid_students)
    print("Average:", average)
    print("Highest scorer:", highest["name"])
    print("Rejected records:", rejected_count)

else:
    print("No valid records found.")
```
