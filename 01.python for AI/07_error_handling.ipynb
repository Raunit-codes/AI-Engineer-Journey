### What Is an Exception?

number = int("hello")

This produces a ValueError.

An exception is an error that occurs while the program is running.

Common examples:

ValueError
TypeError
ZeroDivisionError
FileNotFoundError
KeyError
IndexError

```python
try:
    number = int(input("Enter a number: "))
    print("You entered:", number)

except ValueError:
    print("Please enter a valid integer.")
```

```python
try:
    age = int(input("Enter your age: "))

    if age < 0:
        print("Age cannot be negative.")
    else:
        print(f"You are {age} years old.")

except ValueError:
    print("Age must be a number.")
```

```python
try:
    number = int(input("Enter a number: "))
    result = 100 / number

    print("Result:", result)

except ValueError:
    print("Please enter an integer.")

except ZeroDivisionError:
    print("Number cannot be zero.")
```

```python
try:
    number = int(input("Enter a number: "))
    result = 100 / number

    print(result)

except (ValueError, ZeroDivisionError):
    print("Invalid input.")
```

```python
try:
    number = int(input("Enter a number: "))

except ValueError:
    print("Invalid input.")

else:
    print("Successfully converted:", number)
    
```

## else

The `else` block executes only when the `try` block
finishes successfully.

```python
try:
    number = int(input("Enter a number: "))

except ValueError:
    print("Invalid input.")

else:
    print("Successfully converted:", number)
```

```python
try:
    number = int(input("Enter a number: "))

except ValueError:
    print("Invalid input.")

finally:
    print("This always executes.")
```

try + except + else + finally

```python
try:
    number = int(input("Enter a number: "))
    result = 100 / number

except ValueError:
    print("Input must be an integer.")

except ZeroDivisionError:
    print("Cannot divide by zero.")

else:
    print("Result:", result)

finally:
    print("Operation finished.")
    
```

```python
def calculator():

    try:
        first = float(input("Enter first number: "))
        operator = input("Enter operator (+, -, *, /): ")
        second = float(input("Enter second number: "))

        if operator == "+":
            result = first + second

        elif operator == "-":
            result = first - second

        elif operator == "*":
            result = first * second

        elif operator == "/":
            result = first / second

        else:
            print("Unknown operator.")
            return

        print("Result:", result)

    except ValueError:
        print("Please enter valid numbers.")

    except ZeroDivisionError:
        print("Cannot divide by zero.")


calculator()
```

### raise

deliberately generate an exception.

```python
age = int(input("Enter your age: "))

if age < 0:
    raise ValueError("Age cannot be negative.")

print("Valid age:", age)
```

```python
def calculate_percentage(marks, total):

    if total <= 0:
        raise ValueError("Total marks must be greater than zero.")

    if marks < 0:
        raise ValueError("Marks cannot be negative.")

    if marks > total:
        raise ValueError("Marks cannot exceed total marks.")

    return (marks / total) * 100


print(calculate_percentage(85, 100))
```

```python
calculate_percentage(120, 100)
```

```python
try:
    percentage = calculate_percentage(120, 100)

except ValueError as error:
    print("Validation error:", error)
```

### Custom Exceptions

```python
class InvalidMarksError(Exception):
    pass
```

```python
def validate_marks(marks):

    if marks < 0 or marks > 100:
        raise InvalidMarksError(
            "Marks must be between 0 and 100."
        )

    return True


try:
    marks = float(input("Enter marks: "))

    validate_marks(marks)

    print("Valid marks.")

except InvalidMarksError as error:
    print("Validation error:", error)

except ValueError:
    print("Please enter a number.")
```

### File Errors

```python
try:

    with open("unknown_file.txt", "r") as file:
        content = file.read()

    print(content)

except FileNotFoundError:
    print("The requested file does not exist.")
```

### Dynamic File Reader

```python
filename = input("Enter filename: ")

try:

    with open(filename, "r") as file:
        content = file.read()

    print("\nFile contents:")
    print(content)

except FileNotFoundError:
    print(f"Could not find '{filename}'.")

except PermissionError:
    print("You do not have permission to read this file.")
```

### Handling JSON Errors

```python
import json

filename = input("Enter JSON filename: ")

try:

    with open(filename, "r") as file:
        data = json.load(file)

    print("JSON loaded successfully.")
    print(data)

except FileNotFoundError:
    print("File does not exist.")

except json.JSONDecodeError:
    print("The file contains invalid JSON.")
```

### Handling Dictionary Errors 

```python
student = {
    "name": "Aman",
    "marks": 85
}

try:
    print(student["age"])

except KeyError:
    print("The 'age' field does not exist.")
```

```python
age = student.get("age")

if age is None:
    print("Age information unavailable.")
```

### Handling List Errors

```python
numbers = [10, 20, 30]

try:
    index = int(input("Enter an index: "))
    print(numbers[index])

except ValueError:
    print("Index must be an integer.")

except IndexError:
    print("That index does not exist.")
```

### Dynamic Data validation

```python
def get_positive_number():
    while True:
        try:
            value = float(
                input("Enter a positive number:")
            )
            if value <= 0:
                raise ValueError(
                    "Number must be positive"
                )
            return value
        except ValueError as error:
            print("Invalid input:", error)

number = get_positive_number()
print("Accepted value:",number)
```
