```python
def greet():
    print("Hello, welcome to Python for AI Engineering!")

greet()
```

```python
def greet(name):
    print(f"Hello, {name}!")

greet("Ray")
greet("Alex")
greet("Sam")
```

```python
def introduce(name, age):
    print(f"My name is {name} and I am {age} years old.")

introduce("Ray", 19)
```

```python
def add(a, b):
    return a + b

result = add(10, 20)

print(result)
```

```python
def square(number):
    return number ** 2

result = square(5)

print(result)
print(result + 10)
```

```python
def calculate(a, b):
    addition = a + b
    subtraction = a - b
    multiplication = a * b

    return addition, subtraction, multiplication

result = calculate(10, 5)

print(result)
```

unpacking

```python
addition, subtraction, multiplication = calculate(10, 5)

print(addition)
print(subtraction)
print(multiplication)
```

```python
def analyze_data(numbers):
    total = sum(numbers)
    average = total / len(numbers)
    minimum = min(numbers)
    maximum = max(numbers)

    return total, average, minimum, maximum
```

```python
data = [10, 20, 30, 40, 50]

total, average, minimum, maximum = analyze_data(data)

print("Total:", total)
print("Average:", average)
print("Minimum:", minimum)
print("Maximum:", maximum)
```

default parameters

```python
def greet(name="User"):
    print(f"Hello, {name}!")

greet("Ray")
greet()
```

```python
def profile(name, age, branch):
    print(name, age, branch)

# Positional
profile("Ray", 19, "AI")

# Keyword
profile(
    name="Ray",
    age=19,
    branch="AI"
)
```

*args: allows a function to accept an arbitrary number of positional arguments.

```python
def add_numbers(*numbers):

    total = 0

    for number in numbers:
        total += number

    return total

print(add_numbers(1, 2))
print(add_numbers(1, 2, 3, 4))
print(add_numbers(10, 20, 30, 40, 50))
```

**kwargs: allows arbitrary keyword arguments.

```python
def display_info(**info):

    for key, value in info.items():
        print(f"{key}: {value}")

display_info(
    name="Ray",
    age=19,
    branch="AI",
    college="Engineering College"
)
```

```python
def example(*args, **kwargs):

    print("Arguments:", args)
    print("Keyword arguments:", kwargs)

example(
    10,
    20,
    30,
    name="Ray",
    course="AI"
)
```
