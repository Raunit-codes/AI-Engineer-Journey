### What Is an API?

API = Application Programming Interface

An API allows one software system to communicate with another.

For example:

Python program -> 
    API request -> 
Weather server ->
    API response -> 
Python program

Instead of manually opening a website, our Python program can request
structured data from a server.

Examples of APIs:

- Weather APIs
- Maps APIs
- Payment APIs
- GitHub API
- YouTube API
- AI model APIs
- Stock/financial APIs
- Database APIs

### client and Server

The CLIENT is the application making the request.

The SERVER is the system providing the data/service.

Example:

Python program = Client

Weather API = Server

Python:

"Give me the current weather for Kanpur."

Server:

"Here is the weather data."

### HTTP

Most web APIs communicate using HTTP/HTTPS.

HTTP defines how requests and responses are exchanged.

Common HTTP methods:

GET
POST
PUT
PATCH
DELETE

The most important ones for now:

GET    → retrieve data
POST   → send/create data
PUT    → replace/update data
PATCH  → partially update data
DELETE → remove data

### HTTP Request

A request can contain:

- URL
- HTTP method
- Parameters
- Headers
- Body

Example:

GET https://example.com/users

Conceptually:

Method -> GET

URL: 
https://example.com/users

Parameters->
?city=Kanpur

Headers->
Authentication / content information

Body->
Usually used with POST/PUT/PATCH

```python
import requests

print("requests imported successfully!")
```

```python
url = "https://jsonplaceholder.typicode.com/posts/1"

response = requests.get(url)

print(response)
```

### HTTP Status Codes

The server returns a status code.

Some important ones:

200 → Success

201 → Created

204 → Success, no content

400 → Bad Request

401 → Unauthorized

403 → Forbidden

404 → Not Found

429 → Too Many Requests


500 → Server Error

502 → Bad Gateway

503 → Service Unavailable

2xx → generally successful

4xx → problem with the request/client

5xx → server-side problem

```python
url = "https://jsonplaceholder.typicode.com/posts/1"

response = requests.get(url)

print("Status code:", response.status_code)

if response.status_code == 200:
    print("Request successful!")
else:
    print("Request failed.")
```

```python
print(response.text)
```

### JSON

JSON = JavaScript Object Notation

APIs commonly use JSON to exchange structured data.

Example:

{

JSON looks very similar to Python dictionaries.

JSON:
