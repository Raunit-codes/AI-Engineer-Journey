```python

languages = ["Python", "C++", "Java", "JavaScript"]

print(languages)
print(type(languages))
```

```python
print(languages[0])
print(languages[1])
print(languages[-1])
```

```python
print(languages[0:2])
print(languages[1:3])
print(languages[:2])
print(languages[2:])
print(languages[::-1])
```

```python
languages[1] = "Rust"

print(languages)
```

```python
languages.append("Go")

print(languages)
```

```python
languages.insert(1, "CPP")

print(languages)
```

```python
languages.remove("Java")

print(languages)
```

```python
removed = languages.pop()

print("Removed:", removed)
print("Remaining:", languages)
```

```python
numbers = [10, 20, 30, 40, 50]

print("Length:", len(numbers))
print("Sum:", sum(numbers))
print("Minimum:", min(numbers))
print("Maximum:", max(numbers))
```

```python
print("Python" in languages)
print("Rust" in languages)
```

```python
if "Python" in languages:
    print("Python is available.")
```

```python
marks = [45, 78, 92, 34, 67, 88]

passed = []

for mark in marks:
    if mark >= 50:
        passed.append(mark)

print(passed)
```
